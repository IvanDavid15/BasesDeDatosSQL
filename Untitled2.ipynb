```python
# =========================
# CREAR DATASET STUDENTS
# =========================

students = """
gender,math_score,reading_score,writing_score
female,72,72,74
male,47,57,44
female,90,95,93
male,76,78,75
female,88,95,92
"""

with open("students.csv", "w") as f:
    f.write(students)

# =========================
# CREAR DATASET SALES
# =========================

sales = """
product,sales,profit
Laptop,1200,300
Phone,800,200
Tablet,600,150
Chair,400,80
Desk,700,120
"""

with open("sales_data.csv", "w") as f:
    f.write(sales)

# =========================
# CREAR DATASET HOSPITAL
# =========================

hospital = """
patient,age,temperature
Juan,25,37
Maria,40,38
Carlos,30,39
Ana,50,36
Pedro,60,38
"""

with open("hospital_data.csv", "w") as f:
    f.write(hospital)

# =========================
# CREAR DATASET IOT
# =========================

iot = """
device,temperature,humidity
sensor1,25,60
sensor2,28,55
sensor3,30,70
sensor4,27,65
sensor5,29,68
"""

with open("iot_sensors.csv", "w") as f:
    f.write(iot)

# =========================
# CREAR DATASET BANK
# =========================

bank = """
client,balance,transactions
Juan,5000,20
Maria,7000,35
Carlos,3000,15
Ana,9000,40
Pedro,4500,18
"""

with open("bank_transactions.csv", "w") as f:
    f.write(bank)

print("LOS 5 DATASETS FUERON CREADOS")
```
